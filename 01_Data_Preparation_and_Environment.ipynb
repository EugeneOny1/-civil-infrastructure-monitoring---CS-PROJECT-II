##Dataset Inspection

```python
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"}

def count_images(directory):
    return sum(
        1 for path in Path(directory).rglob("*")
        if path.suffix in IMAGE_EXTENSIONS
    )
```

```python
print("SDNET2018 images:", count_images(SDNET_DIR))
print("RDD2022 images:", count_images(RDD_DIR))
print("Kenyan images:", count_images(KENYAN_DIR))
```

##Inspecting Annotation Formats

```python
def count_files(directory, extensions):
    directory = Path(directory)

    return sum(
        1
        for path in directory.rglob("*")
        if path.suffix.lower() in extensions
    )
```

```python
XML_EXTENSIONS = {".xml"}
YOLO_EXTENSIONS = {".txt"}
JSON_EXTENSIONS = {".json"}

for name, directory in {
    "RDD2022": RDD_DIR,
    "SDNET2018": SDNET_DIR,
    "Kenyan_Data": KENYAN_DIR
}.items():

    print(f"\n{name}")
    print("-" * 30)

    print("Images:", count_images(directory))
    print("XML:", count_files(directory, XML_EXTENSIONS))
    print("YOLO TXT:", count_files(directory, YOLO_EXTENSIONS))
    print("JSON:", count_files(directory, JSON_EXTENSIONS))
```
